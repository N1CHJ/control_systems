```python
import os
import sys

import sympy as sp
from IPython.display import Math, display

sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src'))
from case_studies.common import sym_utils as su
from case_studies.H_hummingbird.generate_linearization import *

su.enable_printing(True)
s = sp.symbols("s")
```

### 1. Longitudinal Transfer Function: P_theta(s) = theta(s) / F_ctrl(s)
From H.4: thetaddot = b_theta * F_ctrl
In Laplace domain: s^2 * theta(s) = b_theta * F_ctrl(s)
P_theta(s) = b_theta / s^2

```python
b_theta_expr = linearized_eom_lon.rhs / F_ctrl
b_theta_expr = b_theta_expr.subs({beta: 0})
b_theta_expr = sp.simplify(b_theta_expr)

P_theta = b_theta_expr / (s**2)

print("\nLongitudinal Transfer Function (Pitch):")
su.printeq("P_theta(s)", P_theta)
```

### 2. Lateral Transfer Functions
From H.4:
phiddot = b_phi * tau
psiddot = a_psi * phi

P_phi(s) = phi(s) / tau(s) = b_phi / s^2
P_psi_phi(s) = psi(s) / phi(s) = a_psi / s^2
P_psi_tau(s) = psi(s) / tau(s) = (a_psi * b_phi) / s^4

```python
# Extract b_phi and a_psi from A_lat and B_lat
# A_lat is 4x4, B_lat is 4x1
# x_lat = [phi, psi, phidot, psidot]
# phiddot is 3rd row (index 2), psiddot is 4th row (index 3)
# u_lat = [tau]

b_phi_expr = B_lat[2, 0]
a_psi_expr = A_lat[3, 0]

P_phi = b_phi_expr / (s**2)
P_psi_phi = a_psi_expr / (s**2)
P_psi_tau = (a_psi_expr * b_phi_expr) / (s**4)

print("\nLateral Transfer Function (Roll):")
su.printeq("P_phi(s)", P_phi)

print("\nLateral Transfer Function (Yaw relative to Roll):")
su.printeq("P_psi_phi(s)", P_psi_phi)

print("\nLateral Transfer Function (Yaw relative to Torque):")
su.printeq("P_psi_tau(s)", P_psi_tau)
```

### 1. Longitudinal State-Space
x_lon = [theta, thetadot]
xdot_lon = [thetadot, thetaddot] = [thetadot, b_theta * F_ctrl]
A_lon = [[0, 1], [0, 0]]
B_lon = [[0], [b_theta]]
C_lon = [[1, 0]]
D_lon = [[0]]

```python
A_lon = sp.Matrix([[0, 1], [0, 0]])
B_lon = sp.Matrix([[0], [b_theta_expr]])
C_lon = sp.Matrix([[1, 0]])
D_lon = sp.Matrix([[0]])

print("\nLongitudinal State-Space Matrices (A, B, C, D):")
su.printeq("A_lon", A_lon)
su.printeq("B_lon", B_lon)
su.printeq("C_lon", C_lon)
su.printeq("D_lon", D_lon)
```

### 2. Lateral State-Space
x_lat = [phi, psi, phidot, psidot]

```python
C_lat = sp.Matrix([[1, 0, 0, 0], [0, 1, 0, 0]]) # Output phi and psi
D_lat = sp.Matrix([[0], [0]])

print("\nLateral State-Space Matrices (A, B, C, D):")
su.printeq("A_lat", A_lat)
su.printeq("B_lat", B_lat)
su.printeq("C_lat", C_lat)
su.printeq("D_lat", D_lat)
```
