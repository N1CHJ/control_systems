```python
#Question 12
import sympy as sp

s, k1, k2 = sp.symbols('s k1 k2')

A = sp.Matrix([[-1, 4], 
               [ 1, 0]])
B = sp.Matrix([[8], 
               ])

K_sym = sp.Matrix([[k1, k2]])

A_cl = sp.Matrix([[-1-k1, 4-k2], 
               [ 1, 0]])
char_eq = (s * sp.eye(2) - A_cl).det()

# -1 +/- 2j
desired_char_eq = (s - (-1 + 2*sp.I)) * (s - (-1 - 2*sp.I))
desired_char_eq = sp.expand(desired_char_eq)
eq1 = sp.Eq(char_eq.coeff(s, 1), desired_char_eq.coeff(s, 1))
eq2 = sp.Eq(char_eq.subs(s, 0), desired_char_eq.subs(s, 0))

# Solve 
gains = sp.solve([eq1, eq2], (k1, k2))
print(f"k1, k2: {gains}")
```

```python
#Question 17:
import sympy as sp

s, K, ki = sp.symbols('s K ki')

desired_char_eq = sp.expand((s - (-1)) * (s - (-2)))

# A_cl = [[K - 5, ki], 
#         [-1,     0]]
actual_char_eq = s**2 + (5 - K)*s + ki
eq1 = sp.Eq(actual_char_eq.coeff(s, 1), desired_char_eq.coeff(s, 1)) # Coefficients
eq2 = sp.Eq(actual_char_eq.subs(s, 0), desired_char_eq.subs(s, 0)) # Constants
gains = sp.solve([eq1, eq2], (K, ki))

print(f"K, ki: {gains}")
```
