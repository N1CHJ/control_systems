```python
# Porblem 14
import sympy as sp

s = sp.symbols('s')
A = sp.Matrix([[0.4, 0.6], 
               [1.0, -0.2]])
B = sp.Matrix([[0.5], 
               [0.0]])
C = sp.Matrix([[0, 1]])
I = sp.eye(2)

#C * (sI - A)^-1 * B + D
sI_minus_A = s * I - A
inverse_term = sI_minus_A.inv()

H_matrix = C * inverse_term * B

transfer_function = sp.simplify(H_matrix[0])
sp.pprint(transfer_function)
```

```python
# Problem 16
s = sp.symbols('s')
denominator = 6*s**2 + 2*s - 9
poles = sp.solve(denominator, s)
for pole in enumerate(poles):
    sp.pprint(pole)
```
